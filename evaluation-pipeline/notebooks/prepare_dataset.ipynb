# Prepare Datasets

Prepare datasets for **LLM-as-a-Judge** experiments using the config-driven
dataset pipeline.

The notebook intentionally contains only high-level workflow logic.
Dataset-specific loading and mapping are handled by the dataset configuration
and pipeline modules.


## 1. Setup

Import only the public functions needed to build, inspect, and save a prepared
dataset.


```python
import logging
from pathlib import Path

from evaluation_pipeline.datasets.dataset_builder import build_dataset
from evaluation_pipeline.datasets.dataset_io import (
    preview_dataset,
    save_dataset_to_file,
    get_prepared_dataset_path,
)

logging.basicConfig(
    level=logging.INFO,
    format="%(levelname)s:%(name)s:%(message)s",
    force=True,
)
```

## 2. Configuration

Choose the dataset and the desired number of output samples.

Supported datasets are defined by YAML files in the dataset configuration
directory. The notebook does not need to know their source schema.


```python
DATASET_ID = "ppe_gpqa"  # e.g. "beavertails", "truthfulqa", "harmbench", "ppe_gpqa"
N_SAMPLES = 100
SEED = 42
OVERWRITE_DATASET = False

OUTPUT_FILE = get_prepared_dataset_path(DATASET_ID)
print(f"Dataset:     {DATASET_ID}")
print(f"Samples:     {N_SAMPLES}")
print(f"Output file: {OUTPUT_FILE}")
```

## 3. Build Dataset

`build_dataset` loads the configured source, maps source rows to the canonical
evaluation schema, assigns metadata, and validates the resulting dataset.


```python
dataset = build_dataset(
    dataset_id=DATASET_ID,
    n_samples=N_SAMPLES,
    seed=SEED,
)
```

## 4. Preview

Inspect the number of samples, label distribution, and one representative
example before saving.


```python
preview_dataset(dataset)
```

## 5. Save Prepared Dataset

Save the validated canonical dataset as JSON. Existing files are preserved
unless `OVERWRITE_DATASET` is enabled.


```python
save_dataset_to_file(
    dataset=dataset,
    path=OUTPUT_FILE,
    overwrite=OVERWRITE_DATASET,
)
```
